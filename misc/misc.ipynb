```python
import torch
import numpy as np

sum_prob = 0
n_classes = 8  # 8
gamma = 0.725  # 0.75, 0.725
start_prob = 30  # 27.5, 30
for i in range(n_classes):
    if i == n_classes-1:
        new_prob = 100 - sum_prob
    else:
        new_prob = start_prob*gamma**i
    sum_prob += new_prob
    print(f"Class {i} prob: {new_prob:.2f}, (cdf: {sum_prob:.2f})")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Parameters for the lognormal distribution
mu = 0       # Mean of the underlying normal distribution
sigma = 0.5  # Standard deviation of the underlying normal distribution

# Generate lognormal data
data = np.random.lognormal(mu, sigma, 1000)

# Create a histogram of the lognormal data
plt.hist(data, bins=30, density=True, alpha=0.6, color='g', label='Histogram')

# Generate points on the x axis
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)

# Calculate the PDF of the lognormal distribution
pdf = stats.lognorm.pdf(x, sigma, scale=np.exp(mu))

# Plot the PDF
plt.plot(x, pdf, 'k', linewidth=2, label='Lognormal PDF')

# Add labels and legend
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.title('Lognormal Distribution')
plt.legend()

# Show the plot
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Example data
data = {
    'preds': [0., 4., 0., 0., 0., 0., 0., 0., 0., 0.],
    'fitness': [0.516, 0.768, 0.982, 0.593, 0.607, 0.472, 0.586, 0.514, 0.897, 0.851],
    'rmse': [0.0565, 0.0294, 0.0149, 0.0531, 0.0512, 0.0583, 0.0533, 0.0564, 0.0307, 0.0303],
    'gts': [0., 3., 0., 0., 0., 0., 0., 1., 0., 0.]
}

df = pd.DataFrame(data)

# Correlation matrix
corr_matrix = df.corr()
print(corr_matrix)

# Heatmap of the correlation matrix
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix Heatmap')
plt.show()

# Scatter plots
sns.pairplot(df)
plt.show()

# Linear regression
X = df[['preds', 'fitness', 'rmse']]
y = df['gts']

model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

# R-squared score
r2 = r2_score(y, y_pred)
print(f'R-squared score: {r2}')

# Plotting actual vs predicted gts
plt.scatter(y, y_pred)
plt.xlabel('Actual gts')
plt.ylabel('Predicted gts')
plt.title('Actual vs Predicted gts')
plt.show()

# Residual plot
residuals = y - y_pred
sns.residplot(x=y_pred, y=residuals, lowess=True)
plt.xlabel('Predicted gts')
plt.ylabel('Residuals')
plt.title('Residuals vs Predicted gts')
plt.show()
```

```python
import torch
import time

def chamfer_distance(pc0, pc1):
    # Compute pairwise distances
    dist_pc0_pc1 = torch.cdist(pc0, pc1)  # Shape: [N, M]

    # Minimum distance from each point in pc0 to any point in pc1
    min_dist_pc0_pc1 = dist_pc0_pc1.min(dim=1).values

    # Minimum distance from each point in pc1 to any point in pc0
    min_dist_pc1_pc0 = dist_pc0_pc1.min(dim=0).values
    del dist_pc0_pc1

    # Chamfer distance is the sum of minimum distances
    chamfer_dist = min_dist_pc0_pc1.mean() + min_dist_pc1_pc0.mean()
    return chamfer_dist

lengths = torch.linspace(100, 30000, 10, dtype=torch.int)
#lengths = [1000, 2000, 3000, 4000, 5000, 6000]
tot_time = 0
for N in lengths:
    pc0 = torch.rand((N, 3), device="cuda")
    pc1 = torch.rand((N, 3), device="cuda")

    start_time = time.time()
    dist = chamfer_distance(pc0, pc1)
    end_time = time.time()
    del pc0, pc1
    print(f"Time taken to compute Sinkhorn distances: {end_time - start_time:.4f} seconds at {N} pts")
```
